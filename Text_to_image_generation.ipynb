```python
from diffusers.utils.import_utils import USE_SAFETENSORS
model_id1 = "dreamlike-art/dreamlike-diffusion-1.0"
model_id2 = "stabilityai/stable-diffusion-xl-base-1.0"
pipe = StableDiffusionXLPipeline.from_pretrained(model_id2, torch_dtype=torch.float16, USE_SAFETENSORS = True)
pipe = pipe.to("cuda")  # PyTorch + Stable Diffusion use
```

```python
pipe.schedular = EulerAncestralDiscreteScheduler.from_config(pipe.scheduler.config)
#pipe.enable_xformers_memory_efficient_attention()
pipe.enable_attention_slicing()
```

```python
# Tumko prompt me style + camera + lighting + rendering + composition sab specify karna padega.
prompt = """dreamlikeart, a grungy woman with rainbow hair, travelling between dimensions, dynamic pose, happy, soft eyes and narrow chin,
extreme bokeh dainty figure, long hair straight down, torn kawaii shirt and baggy jeans
"""
```

```python
import torch, gc
gc.collect()
torch.cuda.empty_cache()
```

```python
image = pipe(
    prompt=prompt,
    negative_prompt="blurry, low quality, bad anatomy, extra limbs, distorted face, weird lighting",
    num_inference_steps = 40,
    guidance_scale = 8,
    height = 1024,
    width = 1024
).images[0]
```

```python
print("[PROMPT]: ",prompt)
plt.imshow(image);
plt.axis("off");
```

```python
def generate_image(pipe, prompt, params):
  img = pipe(prompt, **params).images

  num_images = len(img)
  if num_images > 1:
    fig, ax = plt.subplots(nrows = 1, ncols = num_images)
    for i in range(num_images):
      ax[i].imshow(img[i])
      ax[i].axis("off")

  else:
    fig = plt.figure()
    plt.imshow(img[0])
    plt.axis("off")
  plt.tight_layout()
  plt.show()
```

```python
params = {"num_inference_steps": 100}
```

```python
prompt = """a cinematic wedding scene, the bride and groom dancing together on the dance floor,
full body shot, elegant pose, soft golden lighting, volumetric light rays in the background,
luxury wedding hall, shallow depth of field, bokeh highlights, smooth skin, detailed fabric textures,
8k ultra-detailed photography, high dynamic range, realistic shadows, dramatic yet warm atmosphere,
fine details in the bride’s dress, natural expressions, premium cinematic look, masterpiece"""
```

```python
generate_image(pipe, prompt, params)
```
